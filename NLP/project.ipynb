Text Preprocessing

```python
# We can see that the emotions column is of array type. We will convert it to numerical values using a dictionary mapping.
df['emotions'].unique()
```

```python
# We can also do it manually by creating a dictionary mapping of emotions to numbers.

unique_emotions = df['emotions'].unique()
emotions_numbers = {
    # 'anger': 0, 'fear': 1,'joy': 2,'love': 3,'sadness': 4,'surprise': 5
    }
i = 0
for emotion in unique_emotions:
    emotions_numbers[emotion] = i
    i = i+1

df['emotions'] = df['emotions'].map(emotions_numbers)
```

1. Lowercasing - Convert all text to lowercase

```python
df['text'] = df['text'].str.lower()
```

```python
# We use lambda to make a small function. Words from text will come into x and will be converted to lower case and then returned to x.
df['text'] = df['text'].apply(lambda x:x.lower())
```

```python
df
```

2. Remove Punctuation - Remove symbols like @, #, $, %, ^, &, *

```python
# string is a library used to remove punctuation from text
import string
```

```python
def remove_punctuation(text):
  return text.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punctuation)
```

3. Remove Numbers

```python
def remove_numbers(text):
  new = ""
  for i in text:
    if not i.isdigit():
      new = new + i
    return new
```

```python
df['text'] = df['text'].apply(remove_numbers)
```

6. Remove Emojis and special characters

```python
def remove_emojis(text):
  new = ""
  for i in text:
    if i.isascii():
      new = new + i
      return new
```

```python
df['text'] = df['text'].apply(remove_emojis)
```

7. Remove stopwords - is, am, are, was

```python
# NLTK is a library used for natural language processing. It has many functions to process text data. We will use it to remove stop words from the text data.
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
# punkt is a library used to tokenize text into words
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stopwords = set(set(stopwords.words('english')))
```

```python
stopwords
```

```python
len(stopwords)
```

```python
df.loc[1]['text']
```

```python
def remove_stopwords(text):
  words  = word_tokenize(text)
  cleaned_text = []
  for i in words:
    if i not in stopwords:
      cleaned_text.append(i)
  return " ".join(cleaned_text)
```

```python
df['text'] = df['text'].apply(remove_stopwords)
```

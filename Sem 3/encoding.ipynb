```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('encoding_practice_dataset.csv')
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df['Gender'] = le.fit_transform(df['Gender'])
df
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['School', 'College', 'Postgraduate']])
df['Education_Level'] = oe.fit_transform(df[['Education_Level']])
```

```python
df
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
df['Satisfaction'] = oe.fit_transform(df[['Satisfaction']])
```

```python
df
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
# encoded_data = ohe.fit_transform(df[['City']])
# encoded_df = pd.DataFrame(encoded_data, columns=ohe.get_feature_names_out(['City']))
df['City'] = ohe.fit_transform(df[['City']])
```

```python
df
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
df['Course_Mode'] = ohe.fit_transform(df[['Course_Mode']])
df
```

STANDARD SCALING

```python
df1 = pd.read_csv('scaling_with_outlier_dataset.csv')
df1
```

```python
from sklearn.preprocessing import MinMaxScaler
mn = MinMaxScaler()
df1 = mn.fit_transform(df1)
df1
```

```python
from sklearn.preprocessing import StandardScaler
mn = StandardScaler()
df1 = mn.fit_transform(df1)
df1
```

```python
from sklearn.preprocessing import RobustScaler
mn = RobustScaler()
df1 = mn.fit_transform(df1)
df1
```

```python
df1 = pd.read_csv('scaling_with_outlier_dataset.csv')

scaler = RobustScaler()
df1_scaled = pd.DataFrame(
  scaler.fit_transform(df1),
  columns=df1.columns
)

sns.boxplot(y=df1_scaled['Income'])
plt.show()
```
